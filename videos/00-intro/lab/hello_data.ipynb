Notebook đầu tiên: đọc POS từ Postgres, vẽ doanh thu theo ngày. Chạy **Run All**.

```python
import os
import pandas as pd
from sqlalchemy import create_engine

engine = create_engine(os.environ["POSTGRES_URL"])
sql = """
SELECT (ts AT TIME ZONE 'Asia/Ho_Chi_Minh')::date AS business_date,
       count(*) AS transactions,
       sum(total_amount_vnd) / 1e6 AS revenue_mil_vnd
FROM pos_transactions GROUP BY 1 ORDER BY 1
"""
df = pd.read_sql(sql, engine)
df
```

```python
ax = df.plot.bar(x="business_date", y="revenue_mil_vnd", legend=False, title="Doanh thu theo ngày (triệu VND) — thấy Tết chưa?")
ax.set_xlabel(""); ax.set_ylabel("triệu VND");
```
